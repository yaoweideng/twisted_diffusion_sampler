```python
import pandas as pd
import os

# Corrected paths
test_path = os.path.join('test', 'sc_results', 'all_sc_results.csv')
test2_path = os.path.join('test2', 'sc_results', 'all_sc_results.csv')

def mean_from_csv(csv_path):
    df = pd.read_csv(csv_path)
    mean_rmsd = df['rmsd'].mean()
    mean_motif_rmsd = df['motif_rmsd'].mean()
    return mean_rmsd, mean_motif_rmsd

results = {}
results['test (solvent-exposed offset)'] = mean_from_csv(test_path)
results['test2 (reverted/original)'] = mean_from_csv(test2_path)

for label, (mean_rmsd, mean_motif_rmsd) in results.items():
    print(f'{label}:')
    print(f'  Mean RMSD: {mean_rmsd:.3f}')
    print(f'  Mean motif-RMSD: {mean_motif_rmsd:.3f}\n')
```
